```python
- 객체를 만들면서 필요한 값을 바로 저장하기 위해 사용하는 메서드
- 객체 생송 시 자동으로 호출되어 초기값을 저장하는 역할을 함
- 생성자는 항상 __init__ 라는 이름을 가짐
- 첫번쨰 매개변수는 항상 self
- 생성자는 객체(인스턴스)의 속성을 초기화하는데 사용
- 생성자는 값을 반환하지 않으며, 반환문을 사용하면 안됨!!!
- > 객체가 생성될 떄 자동으로 생성자인 init() 호출됨!!
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def greet(self):
        print(f'Hello, My name is {self.name}')
class Student(Person):
    def study(self):
        print(f'{self.name}학생은 공부를 합니다.')

s = Student('Alice', 20) # 부모 클래스를 모두 물려받기 떄문에 init()동작 -> 값 넣어줘야함
s.greet()
s.study()
```

```python
class Calculator:
    def __init__(self, first, second):
        self.first = first
        self.second = second
    def add(self):
        return self.first + self.second
    def sub(self):
        return self.first - self.second
    def mul(self):
        return self.first * self.second
    # 나눗셈 만들기
    def div(self):
        return self.first / self.second    
```

```python
class FiveCal(Calculator): # 자식 클래스 -> 위의 클래스 상속
    def pow(self):
        return self.first**self.second

a = FiveCal(4,2) # 상속받기에 init()동작 -> 두가지 값을 정확히 전달해야함
print(a.add())
print(a.pow())
```

```python
### 연습문제
- 클래스 C와 클래스 Car를 생성
- 클래스 Car는 클래스 C를 상속받음
- 클래스 C에는 생성자를 통해 데이터를 저장하는 기능 생성 => __init__
- 클래스 Car 는 info_print 메서드를 생성하고 출력기능을 만든다
- 출력되는 내용: 바퀴수 4와 가격 1000원
```

```python
[출력내용]
바퀴수 4
가격 1000
```

```python
class C:
    def __init__(self, m,n): # m,n이 매개변수
        self.wheels = m # 왼쪽이 속성명(변수), 오른쪽이 매개변수
        self.price = n

class Car(C):
    def info_print(self):
        print(f'바퀴수 {self.wheels}')
        print(f'가격 {self.price}') #주소인 self는 필수

c1 = Car(4,1000) # __init__이 자동 동작 =>__init__의 매개변수에 값 저장  , just 저장한거임
c1.info_print()
```
