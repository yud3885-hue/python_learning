## 01 ㅣ 객체지향

```python
01 ㅣ 객체지향
- object oriented programming 의 약자
- 객체지향프로그래밍은 소프트웨어 설계방식 중 하나로, 프로그램을 객체라는 독립적인 단위로 나누어 작성하는 방법
- 객체 = 데이터(속성,변수) + 기능(메서드)을 하나로 묶은 것
- oop는 코드의 재사용성과 유지보수성을 높이는데 도움이 됨
- oop 의 핵심 개념은 클래스,객체
```

```python
# 붕어빵틀 -> 설계도(=클래스) , 이걸로 만든 붕어빵들 -> 객체 / 팥,슈크림 등이 데이터, 붕어빵의 기능이 메서드
```

```python
01) 핵심 개념:
클래스: 객체를 만들기 위한 설계도 (붕어빵틀)
객체: 클래스를 통해 생성된 실체, 결과물(붕어빵 결과물로 만듬)
메서드: 클래스 안에서 정의된 함수
```

```python
02) 객체지향을 사용하는 이유

1. 재사용성
- 한번 만든 클래스를 여러곳에서 재활용 가능
- 불필요한 코드 중복 감소

2. 유지보수성
- 코드가 모듈화되어 있어 수정/보완이 쉬움
- 다른 코드에 영향을 덜 주고 보완이 가능

3. 확장성
- 새로운 기능 추가가 용이(-> 상속, 오버라이딩)
- 기존 코드를 수정하지 않고도 프로그램 확장 가능
```

```python
# 설계도 생성
class Person:
    def setdata(self, name, age, job): #저장기능 # 매개변수 위치 #인수가 있으면 매개변수 쓰기, 인수가 없으면 매개변수 쓰지 않고 걍 self (class에서만,함수는 걍 공백)
        self.name = name # 객체명.속성명 = 매개변수, 속성(= 변수)
        self.age = age 
        self.job = job
    def print_function(self): #출력기능, 위에서 이미 매개변수에 인수를 저장했으므로 가져다쓰면 되기 때문에 따로 또 매개변수 쓰지 않아도 됌
        print(f'이름은 {self.name}이고, 나이는 {self.age}입니다.') # 누구의 속성인지 명확하게 self.속성명으로 적어야함
    def print_job(self): #출력기능
        print(f'직업은 {self.job}입니다.') 

# 객체 생성
p3 = Person()
p3.setdata("James", 28, 'programmer') # p3는 self에 전달, name-> James, age->28, job-> programmer순서대로 각각 매칭
p3.print_function() 
p3.print_job()
```

```python
# 설계도 생성
class Account:
    def setdata(self, owner, balance = 0): # self 에 acc1 매칭, balance 에 setdata의 인수 800 매칭
        self.owenr = owner
        self.balance = balance
    def deposit(self, amount): # self 에 acc1 매칭, amount에 밑에 deposit 기능의 인수 500 받음
        self.balance += amount # amount 500을 받아서 더해서 self.balance에 저장, amount 는 한번 쓰고 사라짐(종료) 
        return self.balance # 총 1300-> 그 값을 return, 반환값 -> acc1.deposit(500)에 전달
    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
        return self.balance
# 객체 생성
acc1 = Account()
acc1.setdata('Alice', 800) #저장 기능, 저장할 값의 인수
amount1 = acc1.deposit(500) # 오른쪽에 return값이 들어있음, 오른쪽에 들고있는 값을 왼쪽에 저장
print(amount1)
```

```python
## 03 ㅣ 사칙연산 계산기 만들기
### 01. 계산기 설계
- 사칙연산이 가능
- 속성 2개: 두개의 값을 저장하기
- 메서드 4개: 덧셈, 뺄셈, 곱셈, 나눗셈
```

```python
class FourCal():
    def setdata(self, first, second):
        self.first = first
        self.second = second
    def add(self):
        self.result = self.first+ self.second #result라는 변수에 담기 , result 라 쓰면 그냥 사라짐, 객체에 저장할라면 self.result라 써야함 (어디에 저장할껀지 명시해야함)
        return self.result
    def mul(self):
        self.result = self.first* self.second
        return self.result
Cal = FourCal()
Cal.setdata(4,2)
print(Cal.add()) #Cal.add() 에 반환값 
print(Cal.mul())
```

```python
class FourCal():
    def setdata(self, first, second):
        self.first = first
        self.second = second
    def add(self):
        self.result = self.first+ self.second #result라는 변수에 담기 , result 라 쓰면 그냥 사라짐, 객체에 저장할라면 self.result라 써야함 (어디에 저장할껀지 명시해야함)
        return self.result
    def mul(self):
        self.result = self.first* self.second
        return self.result
    def sub(self):
        self.result = self.first-self.second
        return self.result
    def div(self):
        self.result = self.first/self.second
        return self.result

# 입력받아서 전달하기
num1 = int(input('첫번째 값을 입력하세요'))
num2 = int(input('두번째 값을 입력하세요'))
Cal = FourCal()
Cal.setdata(num1, num2)
print(Cal.add()) #Cal.add() 에 add 기능의 self.result 반환값 
print(Cal.mul())
print(Cal.sub())
print(Cal.div())
```

```python
## 연습문제 ㅣ
- 이름,나이, 전화번호를 속성으로 가지고 있는 Person 클래스를 이용하여 객체를 생성
- Person 클래스는 info() 메서드를 가지고 있음
- 각각의 속성들을 하나씩 출력해보기
[결과]
James
25
010-0000-1111
```

```python
class Person():
    def info(self, name, age, tel):
        self.name = name
        self.age = age
        self.tel = tel
    def print_info(self):
        print(self.name)
        print(self.age)
        print(self.tel)
m = Person()
m.info('James', 25,'010-0000-1111')
m.print_info()
```

```python
## 연습문제 ㅣ
- CPU, RAM, SSD를 구성요소로 가지고 있는 Computer 클래스를 이용해서 인스턴스를 생성
- Computer  클래스는 각 구성요소의 값을 저장할 수 있는 set_spec() 메서드와
- 구성요소의 값을 출력할 수 있는 hardware_info() 메서드로 구성
```

```python
class Computer:
    def set_spec(self,CPU,RAM,SSD):
        self.CPU = CPU
        self.RAM = RAM
        self.SSD = SSD
    def hardware_info(self):
        print('CPU = ', self.CPU) # or print(f'CPU= {self.CPU}')
        print('RAM = ', self.RAM)
        print('SSD = ',self.SSD)
m = Computer()
m.set_spec('i7', '16GB', '512GB')
m.hardware_info()
```
