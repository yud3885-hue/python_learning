```python
1. 대입연산자
= : 변수에 값을 할당하는데 사용하는 연산자
-> 오른쪽에 값을 대입하는것 뿐만아니라, 오른쪽 연산의 결과를 왼쪽 변수에 대입하는 데에도 사용 가능
```

```python
a = 3
b = 10+2
print(a)
print(b)
```

```python
2. 복합대입연산자
- 대입연산자와 함께 산술연산자를 결합하여 다양한 연산을 수행
- 이러한 연산을 더욱 간단하게 표현할 수 있는 연산자가 복합대입연산자임
```

```python
+= : a += 2 -> 변수 a에 2를 더한 값을 a에 저장
-= :  a -= 2 -> 변수 a에 2를 뺀 값을 a에 저장
*= :  a *= 2 -> 변수 a에 2를 곱한 값을 a에 저장
**= :  a **= 2 -> 변수 a에 2를 거듭제곱한 값을 a에 저장
/= :  a /= 2 -> 변수 a에 2로 나눈 값을 a에 저장
//= :  a //= 2 -> 변수 a에 2로 나눈 몫을 a에 저장
%= :  a %= 2 -> 변수 a에 2로 나눈 나머지를 a에 저장
```

## 04ㅣ 논리 연산자 

```python
## 04ㅣ 논리 연산자 (조건 체크할때 사용)
- 논리적인 연산을 수행하여 참인지 거짓인지를 판단하는 연산자
- 두개 이상의 논리값을 조합하여 새로운 논리값을 만들 수 있음
 => True, False
```

```python
and : 연산자를 기준으로 왼쪽과 오른쪽 값이 모두 True 일때만 True 를 반환
      나머지는 모두 False를 반환
or : 연산자를 기준으로 왼쪽과 오른쪽 값 중 하나라도 True 이면 True를 반환하고, 둘 다 False 일때만 False를 반환
not: 뒤에 따라오는 논리값이 True 이면 False를, False이면 True를 반환
```

```python
01) and:
```

```python
print(True and True)
```

```python
print(False and True)
print(True and False)
print(False and False)
```

```python
02) or: a와 b 중 하나라도 True 이면 True, 아니면 False
```

```python
print(True or True)
print(True or False)
print(False or True)
```

```python
print(False or False)
```

```python
[작성법]
if 1조건문:
    1조건문의 결과가 True일때 수행할 문장
elif 2조건문:
    1조건문의 결과가 False 이고, 2조건문의 결과가 True 일 때 수행할 문장
elif 3조건문:
    1,2조건문의 결과가 False 이고, 3조건문의 결과가 True 일 때 수행할 문장
else:
    1,2,3 조건문의 결과가 False일 때 수행할 문장
```
