```python
money = int(input('돈을 넣어주세요: '))
while True:
    print(f'=== 자판기 ===')
    print(f'1. 콜라 (1000원)')
    print(f'2. 사이다 (1200원)')
    print(f'3. 커피 (800원)')
    print(f'0. 종료')
    choice = int(input('메뉴 선택: '))
    if choice == 1:
        if money >= 1000:
            money -= 1000
            print('콜라를 구매했습니다.')
        else:
            print('잔액이 부족합니다.')
    elif choice == 2:
        if money >= 1200:
            money -= 1200
            print('사이다를 구매했습니다.')
        else:
            print('잔액이 부족합니다.')
    elif choice == 3:
        if money >= 800:
            money -= 800
            print('커피를 구매했습니다.')
        else:
            print('잔액이 부족합니다.')
    elif choice == 0:
        print('자판기를 종료합니다')
        break
    else:
        print('잘못된 선택입니다.')
        
    print(f'현재 잔액: {money} 원')    
            
```
