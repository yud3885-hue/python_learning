```python
* 사람이 읽을 수 있는 시간으로 변경하기 !!!
- datetime모듈을 사용
-> fromtimestamp() 메서드
-> 초단위 숫자를 사람이 읽을 수 있도록 변환
-> strftime(): string format time의 약자 ,datetime객체를 문자열로 포맷팅 (변환)
```

```python
## 날짜 시간 관련 포맷 코드 
%Y | 연도(4자리)
%y | 연도(2자리)
%m | 월(2자리)
%d | 일(2자리)
%H | 시(24시간제, 2자리)
%M | 분(2자리)
%S | 초(2자리)
```

```python
datetime모듈안에
datetime클래스가 있음
fromtimestamp()는 그 클래스의 메서드
```

```python
import time
from datetime import datetime
#datetime.py          class datetime

#현재시간 
now = time.time()
print("현재시간: ", now)
# datetime객체로 변환
dt = datetime.fromtimestamp(now)
print("datetime 객체: ", dt)
#포맷팅된 문자열로 출력
formatted = dt.strftime("%Y.%m.%d")
t = dt.strftime("%H시")
print("형식화된 시간: ", formatted)
print(t)
```

```python
### [1] 파일에 내용 쓰고 읽는 프로그램 만들기
- 사용자가 입력한 내용을 텍스트 파일에 저장하고, 
  저장된 파일 내용을 불러와 화면에 출력하는 간단한 프로그램을 만드세요.

- 파일에 한 줄 쓰고, 읽어서 출력하는 기능만 있는 예제
```

```python
# 파일에 쓰기
def write_file():
    text = input("파일에 쓸 내용을 입력하세요:")
    with open("data.txt", "w", encoding = "utf-8") as f: 
        f.write(text)    
    print("파일에 저장했어요!")
# 파일 읽기
def read_file():
    try:
        with open("data.txt", "r", encoding = "utf-8") as f:
            content = f.read()
        print("파일내용: ")
        print(content)
    except FileNotFoundError:
        print("파일이 없어요")
# 실행
while True:
    print("1. 쓰기  2. 읽기  3. 종료")
    choice = int(input("선택 : "))
    if choice ==1:
        write_file()
    elif choice ==2:
        read_file()
    elif choice == 3:
        print("끝!")
        break
    else:
        print("1,2,3 중 선택하세요. ")
```
