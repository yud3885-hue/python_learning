## 01 ㅣ 크롤링
## 01) 크롤링이란?

```python
- 여러 웹사이트를 자동으로 돌아다니며 필요한 데이터를 수집하는 기술
```

```python
02) 크롤링을 사용하는 이유
- 반복작업 자동화
- 실시간 데이터 수집 가능
- 데이터 분석을 위한 수집단계
```

```python
03) 크롤링 vs 스크래핑
1. 크롤링(crawling)
여러 페이지를 돌아다니며 데이터를 수집하는 과정
2. 스크래핑(scraping)
특정 페이지에서 원하는 정보만 추출하는 작업
```

```python
크롤링: 도서관을 돌아다니는것
스크래핑: 책에서 필요한 문장만 복사하는 것
```

```python
04) 데이터 수집 자동화란?
사람이 반복적으로 수행하는 데이터 수집 작업을 프로그램이 대신 수행하도록 만드는것
```

```python
- 매일 아침 환율을 확인한다
- 매일 뉴스 제목을 저장한다
-> 이런 작업들을 사람이 직접 확인하는 것이 아닌, 프로그램이 대신하도록 만드는 것이 자동화
```

```python
05) 크롤링의 기본흐름
1. 웹페이지에 요청(Request)
2. 서버가 응답(Response)
3. HTML 또는 JSON 데이터 수신
4. 원하는 데이터 추출
5. 파일 또는 DB에 저장
```

```python
HTML
- 웹페이지를 구성하는 마크업 언어
- 태그(Tag)로 구성이 되어 있으며, 웹페이지의 구조를 표현하는 언어
```

```python
<태그> 내용 </태그>
```

```python
<html>
    <body>
        <h1>제목</h1>
        <p>제목</p>
    </body>
</html>
```

```python
06) 크롤링의 3가지 방식

1. HTML을 직접가져오는 방식
-> 웹페이지의 소스코드(HTML)에서 데이터 추출
-> 페이지 요청시 데이터가 바로 보이는 경우

2. 서버에서 데이터를 직접 요청하는 방식 -> API 활용
-> 웹사이트가 내부적으로 사용하는 데이터 요청 경로를 이용
- 구조화된 데이터를 받아서 사용

3. 브라우저를 직접조작하는 방식 -> selenium 
- 사람이 웹사이트를 사용하는 것처럼 자동으로 조작
- 클릭, 로그인, 스크롤 등 가능
```

```python
07) 정적 페이지와 동적페이지
1. 정적페이지: 요청하면 HTML 안에 데이터가 모두 들어 있음
            => HTML 을 가져와서 분석하고, 데이터 추출
2. 동적페이지: JavaScript 실행 후 데이터 생성
            => 먼저 API 가 있는지 확인 후 사용하기
            => API 없으면 브라우저 자동화 사용하기
```

```python
## 02 ㅣ 정적 크롤링
01) 정적크롤링이란?
- 페이지 요청시 서버가 보내주는 HTML 문서 안에 이미 데이터가 포함되어 있는 경우
그 HTML을 분석하여 데이터를 추출하는 방식

02) 정적페이지의 특징
- 페이지 소스보기에서 원하는 데이터가 보인다!
- JavaScript  실행 없이 데이터 확인 가능
- 구조가 비교적 단순하다

F12
or
마우스우클릭 -> 검사
```

```python
06) BeautifulSoup으로 파싱(분석)
- 태그선택방법(find, find_all, select)

- find(): 조건에 맞는 첫번째 태그 찾기
- find_all(): 조건에 맞는 모든 태그 리스트
- select(): CSS 선택자로 여러개 선택
- class / id 선택 방법
```

```python
1. 기본 준비하기
```

```python
from bs4 import BeautifulSoup
html = """
<html>
    <body>
        <h1 id = 'title'>웹크롤링 수업</h1>
        <p class = 'content'>
        <p class = 'content'>
        <a>
    </body>   
</html>
"""

soup = BeautifulSoup(html, 'html.parser')
```

```python
2. find() -> 태그 1개 선택
```

```python
title = soup.find('h1') # h1 태그 선택해서 파싱해줘
print(title)
print(tile.text)
```

```python
3. find_all() -> 태그 여러개 선택
- 조건에 맞는 모든 태그를 리스트로 반환
```

```python
contents = soup.find_all('p')
print(contents)
```

```python
for c in contents: # 여러개가 리스트에 있으므로 반복해서 꺼내기 가능
    print(c)
    print(c.text)
```

```python
4. class 선택 방법
```

```python
class란?
HTML 에서 여러개의 태그를 같은 그룹으로 묶을 때 사용하는 속성
같은 class 이름을 여러태그가 가질 수 있음
그래서 보통 find_all() 과 함께 많이 사용
```

```python
ex)
<p class = 'content'>첫번째문장</p>
<p class = 'content'>두번째문장</p>
<p class = 'content'>세번째문장</p>

# 내용만 다른 여러개의 태그를 같은 클래스로 묶음
```

```python
c = soup.find('p', class_= 'content') # p태그 찾아줘, class의 이름이 content인 곳에서
print(c.text)
```

```python
contents = soup.find_all('p', class_ = 'content')
print(contents)
```

```python
for c in contents:
    print(c.text)
```

```python
5. id 선택방법
- HTML 에서 특정 요소를 구분하기 위해 붙이는 식별자
```

```python
6. select() -> CSS 식별자 방식
- CSS 문법 그대로 사용 가능
- 클래스 앞에는 .
- 아이디 앞에는 #
```

```python
print(soup.select('p'))
```
