```python
- 특정 작업을 수행하는 코드 블록을 의미
- 프로그래밍에서 함수는 코드의 재사용성을 높임
- 프로그램 더 이해하기 쉽고 관리하기 쉬움
-> 함수를 사용 동일한 코드를 반복적으로 작성하지 않고도 필요할 때마다 해당 기능을 호출하여 사용 가능
```

```python
01) 함수 용어 정리
- 내장함수: 이미 만들어져 있어서 사용법대로 사용하면 되는 함수
- 사용자 함수: 사용자가 직접 만든 함수
- 함수 정의: 사용자 함수를 새롭게 만드는 것
- 인수: 사용자 함수에 전달할 입력(argument)
- 매개변수: 인수를 받아서 저장하는 변수 (parameter)
- 반환값: 사용자 함수의 출력(리턴값)
- 함수 호출: 만들어진 사용자 함수를 실제로 사용하는 것
```

```python
02) 함수를 사용하는 이유
- 중복코드 제거: 동일한 기능을 여러번 작성할 필요 없이, 한번 정의한 함수를 여러곳에서 호출해서 사용 가능
- 디버깅 및 유지보수 용이: 함수 안에 문제가 생기면 그 부분만 수정하면 되므로, 디버깅, 유지보수가 쉬워짐
* 디버깅: 오류(버그)를 찾고 원인을 분석하여 수정하는 과정
```

```python
06) 가변 매개변수
- 함수로 전달해야하는 인수의 개수가 정해지지 않았을 때 사용
- 키워드는 애스터리스크(*)로 매개변수 앞에 *를 붙이면 곧바로 가변 매개변수가 됨.
- 전달되는 모든 인수를 하나의 튜플로 만들어줌
- *args는 항상 마지막 매개변수 자리에 와야함
```

```python
01) 매개변수가 *args 하나일 때

[작성법]
def 함수명(*args)

*args의 이름을 변경해도 상관 없음 '*'를 붙여주면 됨
```

```python
def showNumber(*args): # 튜플의 형태로 저장
    print('Args', args)
    print('Args type: ', type(args) )
showNumber(1,2,3,4,5)
```

```python
2) 매개변수가 2개 이상인 경우
- 가변 매개변수는 가장 마지막 위치하도록 함
```

```python
def greet(greeting, *args):
    print(greeting)
    print(args)
    for name in args:
        print(f'{greeting}, {name}!')
greet('hello', 'Jane','Alice','James') # 첫번째 값만 GREETING에 할당, 나머지 값들 모두를 *ARGS에 할당
```

## 04 ㅣ 지역변수와 전역변수

```python
- 변수의 종류에 따라 사용범위와 라이프사이클이 달라짐
 1. 지역변수 (Local Variable)
- 함수 내부에서 선언된 변수, 함수가 실행될 때 생성되고, 함수가 종료되면 사멸
 2. 전역변수(Global Variable)
- 함수 외부에서 선언된 변수, 프로그램 전체에서 접근할 수 있음
- 프로그램이 종료될 때까지 메모리에 존재
```

```python
var = 'global' # 전역변수
def function():
    var = 'local' # 지역변수-> 함수 호출해야만 볼 수 있고 함수 종료되면 메모리에서 사라짐
    print(var)
function()
print(var)
```

```python
## 연습문제 02 | 모든 입력의 평균값 구하기
- 주어진 모든 수의 평균값을 계산해 주는 함수를 작성
- 주어진 수의 개수는 정해져 있지 않다
# TEST코드
```

```python
def calculate_average(*args):
    if not args:
        return None
    total = 0 
    count = 0
    for number in args:
        total += number
        count += 1
    avg = total / count
    return avg
print(calculate_average(1, 2, 3, 4, 5))  # 출력: 3.0
print(calculate_average(10, 20, 30))    # 출력: 20.0
print(calculate_average())              # 출력: None (입력이 없는 경우)
```

```python
def reverse_text(text):#"python"
    result = ""
    for ch in text:
        result = ch+result
        print(result)
    
    return result

print(reverse_text("python"))   # nohtyp
# print(reverse_text("apple"))    # elppa
```
