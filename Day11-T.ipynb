```python
03. lambda와 정렬
key매개변수사용 -> 정렬기준을 지정할 수 있는 옵션
```

```python
[작성법]
변수명.sort(key = lambda x: len(x))
```

```python
-> 익명함수를 사용하여, 문자열길이를 기준으로 정렬
-> 리스트의 각 원소 x에 대해 len(x)값을 비교하여 정렬
-> 정렬할 때 파이썬은 리스트의 각 원소를 꺼내서 key전달
->  그때 원소가 x라는 이름으로 lambda함수에 들어감
```

```python
#오름차순
words = ['apple', 'banana', 'pineapple', 'kiwi']
words.sort(key = lambda x: len(x))
print(words)
```

```python
#내림차순
words = ['apple', 'banana', 'pineapple', 'kiwi']
words.sort(key = lambda x: len(x), reverse = True)
print(words)
```

```python
[작성법]
map(함수명, 반복가능객체)
-> 함수: 각 요소에 적용할 함수 
-> 반복가능객체: 리스트, 튜플, 문자열 등 반복 가능한 객체 
```

```python
#1. 기본사용법 
number = [1,2,3,4,5]
def mul(n):
    return n*2

#map함수 사용
result =map(mul, number)
print(result)
#결과를 리스트로 변환 
result_list = list(result)
print(result_list)
```

```python
2. [1,2,3,4,5]에 5를 더해서 반환하기 
```

```python
#중간의 띄어진 줄제거하는 방법
with open("data3.txt", "r", encoding = "utf-8") as f:
    line1 = f.readline()
    line2 = f.readline()
    line3 = f.readline()
    print("첫 줄: ", line1, end = "")
    print("둘째줄: ", line2, end = "")
    print("셋째줄: ", line3)
```

```python
#중간의 띄어진 줄제거하는 방법
with open("data3.txt", "r", encoding = "utf-8") as f:
    line1 = f.readline()
    line2 = f.readline()
    line3 = f.readline()
    print("첫 줄: ", line1.strip())
    print("둘째줄: ", line2.strip())
    print("셋째줄: ", line3)
    #"안녕하세요!두번째줄입니다.세번째줄입니다"
```

```python
02. try - except - finally문
- finally 절은 try문을 수행 도중에 예외 발생 여부에 상관없이 항상 수행 
- 보통 finally절은 사용한 리소스를 close해야할 때 많이 사용 
ex. 파일을 열었다면 닫아야함!!!
```

```python
try: 
    f = open("text.txt", 'w', encoding = 'utf-8')
    #무언가를 수행함!!
except Exception as e:
     print(e)
finally: 
    f.close() #중간에 오류가 발생했더라도 무조건 실행
```

```python
-> 파일을 쓰기 모드로 연 후 예외 발생 여부에 상관 없이 항상 파일을 닫아주려면 
try - finally문을 사용
```

```python
[작성법]
try: 
    예외가 발생할 수 있는 코드블록
except: 
    예외가 발생했을 때 수행되는 코드블록 
finally: 
    예외 발생과 상관 없이 항상 실행되는 코드 블록
```

```python
try: 
    f = open("day01.txt", 'w', encoding = 'utf-8')
    f.write("day01.txt.입니다.")
except EOFError as e: 
    print(e)
finally: 
    f.close() #파일닫기 
print(f.closed)#파일 close상태 확인 : Ture, False반환
```

```python
[작성법]
try: 
    예외가 발생할 수 있는 코드블록
except: 
    예외가 발생했을 때 수행되는 코드블록
else: 
    정상적으로 실행되면 수행되는 코드블록 #오류가 없을 경우에만 수행
```

```python
try: 
    age = int(input('나이를 입력하세요: '))
except:
    print('입력이 정확하지 않습니다.')
else: 
    if age <= 18:
        print('미성년자에게는 판매금지입니다.')
    else:
        print("구매가능합니다")
```

```python
이름을 입력하세요: 이현2
사람의 이름에는 숫자가 들어갈 수 없습니다.
프로그램을 종료합니다
[정상동작]
이름을 입력하세요: 이현
이현님, 안녕하세요!
프로그램을 종료합니다
```

```python
class NameError(Exception):
    def __init__(self):
        super().__init__("사람의 이름에는 숫자가 들어갈 수 없습니다.")
def is_there_digit(name):
    for temp in name:#l, h, 2
        if temp.isdigit():
            return True
    return False
try: 
    name = input("이름을 입력하세요")
    if is_there_digit(name):#숫자가 있다면 -> if True
        raise NameError
except NameError as e: 
    print(e)
else: 
    print(name, "님, 안녕하세요!")
finally: 
    print("프로그램을 종료합니다.")
```
