01) 자료구조란?

```python
## 01 ㅣ 자료구조
01) 자료구조란?
- 컴퓨터 프로그램 안에서 데이터를 체계적으로 저장하고 관리하는 방식
- '데이터를 어떻게 구조화할 것인가'에 대한 고민
-> 저장공간이 제한되어 있기 때문

02) 자료구조가 왜 필요한가?
- 프로그래밍 다루는 거의 모든 문제는 '데이터를 처리하는 문제'이다
- 데이터를 무작정 저장하면, 원하는 정보를 찾는데 시간이 오래 걸리거나, 추가/삭제 작업이 복잡해진다.
=> 데이터를 효율적으로 저장,탐색,삭제하기 위해 자료구조가 반드시 필요
```

## 02 ㅣ 알고리즘

```python
## 02 ㅣ 알고리즘
01) 알고리즘이란?
- 컴퓨터로 문제를 풀기위한 단계적인 절차,
- 정확히 말하면 문제와 컴퓨터가 주어진 상태에서 문제를 해결하는 방법을 정밀하게
장치가 이해가능한 언어로 기술한것

 02) 효율적인 알고리즘이란?
- 알고리즘이 시작하여 결과가 나올 때까지의 수행시간이 짧으면서 컴퓨터 내에 있는
메모리와 같은 자원을 덜 사용하는 알고리즘
```

```python
03) 수행시간 측정방법
```

```python
import time
def example_algorithm(n):
    total = 0
    for i in range(n):
        total+= i
    return total
start_time = time.time() # 시작시간 기록
result = example_algorithm(1000000)
end_time = time.time() # 종료시간 기록
ex_time = end_time - start_time # 수행시간 기록
print(f'알고리즘 실행시간 : {ex_time: .6f}초')
```

```python
04) 알고리즘 분석방법
01. 시간복잡도(time complexity)
- 알고리즘의 절대적인 수행시간을 나타내는 것이 아님
- 알고리즘을 이루고 있는 연산들이 몇번이나 수행되는지를 숫자로 표시
- 연산에는 덧셈, 곱셈과 같은 산술연산, 대입연산, 비교연산 등이 있음
-> 수행하는 연산의 개수를 계산하여 두개의 알고리즘을 비교할 수 있음
```

```python
05) 점근적 표기법
1. Big-omega - 최선의 경우
-의미: 알고리즘이 가장 빨리 끝나는 경우의 수행시간
- 예: 리스트가 이미 정렬되어 있는경우, 한번만 순회하고 정렬이 끝남

2. Big-Theta - 평균적인 경우
-의미: 일반적으로 예상되는 수행 시간
- 예: 정렬상태가 섞여 있어, 대부분의 요소를 비교하고 일부 교환이 발생

3. Big-O - 최악의 경우
-의미: 알고리즘이 가장 오래걸리는 경우의 수행시간
-예: 리스트가 완전히 역순으로 섞여 있어서, 모든 요소를 서로 비교하고 교환해야하는 경우

* 왜 Big-O 표기법을 가장 많이 사용할까? 
- 최악의 상황을 기준으로 성능을 보장할 수 있음
- 실제 서비스나 시험에서는 '이 알고리즘이 절대 이보다 더 느려지지는 않는다'는 보장이 중요함
```

```python
def print_all(arr): # arr = [1,2,3,4]
    for num in arr:
        print(num)
arr = [1,2,3,4]
print_all(arr)
```

```python
O(n^2) - 이차시간
- 입력크기에ㅐ 대한 시간 복잡도가 제곱으로 증가하는 경우,
주로 중첩루프(이중for문)에서 발생
```

```python
def compare_all(arr):
    for i in range(len(arr)):
        for j in range(len(arr)):
            print(arr[i], arr[j])
arr = [1,2,3,4]
compare_all(arr)

# 4*4 = 16
```

```python
O(log n) - 로그시간
-> 입력크기가 커질수록 실행시간이 로그 함수처럼 느리게 증가하는 경우
-> 데이터 양이 증가해도 성능이 크게 차이 나지 않음

ex) 이진탐색(Binary Search)
num = [1,2,3,4,5,6,7,8,9]
key = 6 (찾는데이터) 
# middle 값을 기준으로 찾는 데이터를 비교 (middle 기준으로 관련 없는 쪽은 제외)
=> 한번 처리할 때마다 남은 처리량이 절반 씩 줄어드는 것
```

## 02 ㅣ Stack 스택

```python
## 02 ㅣ Stack 스택
01) 스택의 특징
1. 스택은 위에서부터 데이터를 삽입하고 위에서 부터 꺼냄
# 컵에 모래가 한칸씩 쌓인다고 비주얼화
2. LIFO(후입선출) -> 들어간 순서의 역순으로 꺼내짐, 마지막에 들어간 데이터가 제일 먼저 꺼내짐
3. 사이즈가 정해저 있으므로, 값이 가득차 있는지, 비어 있는지 확인을 해야함
4. 스택의 삽입: 가득차 있다면 삽입 불가능
5. 스택의 삭제: 비어 있다면 삭제 불가능
```

```python
07) 예외처리
1. 스택 오버플로우 처리(push())
- 스택이 가득 찼을 때 push()를 시도하면 StackOverflowError  예외가 발생
2. 스택 언더플로우 처리(pop())
- 스택이 비어있을 때 pop()을 시도하면 StackUnderflowError 예외가 발생
```

```python
08) 스택구현하기
```

```python
class StackOverflowError(Exception):
    pass
class StackUnderflowError(Exception):
    pass
class Stack:
    def __init__(self, limit):
        self.stack = []
        self.limit = limit
    def is_empty(self):
        return len(self.stack) == 0 # True False # self에는 객체 이름 들어감 비어있다는 뜻 
    def is_full(self):
        return len(self.stack) == self.limit # True False
    def push(self, item):
        if self.is_full(): # 메서드 #True -> 가득찬 상태
            raise StackOverflowError('Stack Overflow Error')
        self.stack.append(item)
    def pop(self):
        if self.is_empty(): # 메서드 #True -> 비어있는 상태 # 괄호 없으면 속성값, () 있으면 메서드
            raise StackUnderflowError('Stack Underflow Error')
        self.stack.pop()
    def __str__(self):
        return f'Stack: {self.stack}'
try:
    s= Stack(limit = 5)
    s.push(1)
    print(s)
    s.push(2)
    print(s)
    s.push(3)
    print(s)
    s.push(4)
    print(s)
    s.push(5)
    print(s)
    s.push(6)
    print(s) #  다시
    s.pop()
    print(s)
    s.pop()
    print(s)
    s.pop()
    print(s)
    s.pop()
    print(s)
    s.pop()
    print(s)
    s.pop()

except StackOverflowError as e:
    print(e)
except StackUnderflowError as e:
    print(e)
```

## 03 ㅣ 큐(Queue)

```python
01) 큐의 특징
- 데이터를 한쪽 끝에서 넣고, 반대쪽 끝에서 꺼내는 선형자료 구조 
                (Enqueue),                       (Dequeue)
# 뒤에서 넣으면 앞쪽부터 꺼낸다 (가로통에서 넣는걸로 생각)
- 선입선출 구죄 FIFO (First In First Out)
               먼저 들어간 데이터가 먼저 나감
```

```python
02) 큐의 주요 연산 메서드
- enqueue(): 뒤쪽에 요소를 추가
- dequeue(): 앞쪽에서 요소를 제거하고 반환
- is_empty(): 큐가 비어 있는지 확인
- is_full(): 큐가 가득 찼는지 호가인
```

```python
03) 사용예:
데이터 처리에서 순서를 유지하면서 작업을 진행해야하는 경우에 주로
[ex] 작업대기열: 작업들이 순서대로 처리
                프린터 작업관리
```

```python
04) 예외처리
1. QueueOverflowError
- 큐가 가득 찼을 때 enqueue()를 시도하면 QueueOverflowError 예외 발생
2. QueueUnderflowError
- 큐가 비어있을 때 dequeue()를 시도하면 QueueUnderflowError 예외 발생
```

```python
05) 큐 구현하기
```

```python
class QueueOverflowError(Exception): # 예외처리
    pass

class QueueUnderflowError(Exception):
    pass

class Queue:
    def __init__(self, limit=5): # 기본값 세팅
        self.queue = []
        self.limit = limit
    def is_empty(self):
        return len(self.queue) == 0 # True False
    def is_full(self):
        return len(self.queue) == self.limit # True False
    def enqueue(self, item):
        if self.is_full():
            raise QueueOverflowError('Queue Overflow Error') # 예외처리할 때 다음 문장도 전달
        self.queue.append(item)
    def dequeue(self):
        if self.is_empty():
            raise QueueUnderflowError('Queue Underflow Error')
        return self.queue.pop(0)  # 0번 인덱스를 꺼낸다 (꺼낸값을 반환)
    def print(self):
        print(f'Queue: {self.queue}')
try:
    q = Queue(limit=5)
# 여기 다시 적기
```

```python
### 문제 풀이 방법 (스택 활용)
- 문자를 하나씩 스택에 push
- 스택에서 하나씩 pop해서 새로운 문자열에 추가
- 스택은 LIFO(후입선출) -> 들어간 순서의 역순으로 꺼내짐
```

```python
def reverse_with_stack(word): # 'hello'
    stack = [] # 스택 간단하게 쓰기
    # stack = ['h', 'e', 'l', 'l', 'o']
    for i in word:
        stack.append(i)
    reversed_word = ''
    while stack: # stack에 값 없으면 종료
        reversed_word += stack.pop()
    return reversed_word

print(reverse_with_stack('hello'))# 출력: olleh
```
