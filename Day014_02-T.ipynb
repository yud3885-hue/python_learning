```python
3. 요청 헤더(봇 차단 방지용)
헤더란? 
- User-Agent와 같은 HTTP요청을 보낼 때 함께 보내는 정보 
-> 즉, "나 이런환경에서 이걸 요청해! 브라우저는 이거고, 언어는 이거, 쿠키는 이런거 사용"
```

```python
user-Agent 확인방법 
1. F12
2. 개발자 도구 - Network 클릭
```

```python
headers = {"User-Agent": "Mozilla/5.0"}
# 크롬, 파이어폭스 등 브라우저인척하기 위한 값
# 이 값을 설정하지 않으면 봇으로 인식되어 차단될 수 있음
```

```python
4. GET요청 :"이 주소의 정보를 줘" 라고 요청하는것!
```

```python
- requests.get()은 웹페이지에 정보를 요청하는 메서드사용 
- header를 같이 보내면 브라우저처럼 행동하여 차단 방지 
- 응답이 성공하면 HTML 을 받아 올 수 있음
```

```python
response = requests.get(url, headers = headers)
```

```python
5. 응답이 성공적일 때 
- 기사제목 가져오기 
```

```python
soup.select()
- CSS선택자로 HTML요소를 리스트 형태로 반환 
- 조건에 맞는 요소가 여러개면 전부 가져옴
```

```python
[ex]
<div class = "sa_text">
    <strong class = "sa_text_strong">헤드라인1 </strong>
</div>
<div class = "sa_text">
    <strong class = "sa_text_strong">헤드라인2 </strong>
</div>
```

```python
from bs4 import BeautifulSoup
html= """
<div class = "sa_text">
    <strong class = "sa_text_strong">헤드라인1 </strong>
</div>
<div class = "sa_text">
    <strong class = "sa_text_strong">헤드라인2 </strong>
</div>
"""
soup = BeautifulSoup(html, "html.parser")
result = soup.select("div.sa_text strong.sa_text_strong")
print(result)
print(result[0])
for content in result:
    print(content.text)
```
