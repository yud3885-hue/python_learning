## 01 ㅣ연결리스트(Lincked List)

```python
- 연결리스트는 데이터를 차례대로 연결해 저장하는 구조
- 데이터를 연속된 메모리 공간에 저장하지 않고, 각 데이터가 다음 데이터의 위치를
가리키는 형태로 구성된 선형 자료 구조
- 각 데이터는 단순히 저장되는 것이 아니라, 
[자기자신 + 다음 데이터를 가리키는 주소(포인터)]를 함께 가지고 있음
* 포인터: 메모리 주소를 저장하는 변수
- 각 데이터 단위를 노드라고함 -> 노드는 데이터와 함께 다음노드 주소를 저장
```

```python
01) 연결리스트가 실행활에서 쓰이는 곳
1. 웹브라우저의 '뒤로가기/ 앞으로가기'
-[네이버]->[구글]->[다음]
2. 음악 재생플레이리스트
```

```python
02) 연결리스트의 주요 장점
1. 크기 제한 없음
2. 삽입,삭제가 빠름
배열: [0,1,2,3,4,5]
연결리스트: [0]-[1]-[2]-[3]-[4]-[5]
-> 배열에서 3 삭제시 뒤에 있는 데이터가 앞으로 이동해야함
-> 연결리스트에서는 삭제 후 연결만 바꾸어 주면 됨!
=> 그래서 삽입 삭제가 빠름
```

```python
03) 연결리스트를 사용하면 좋은 경우
1. 데이터 양이 자주 바뀌는 경우
2. 중간 삽입/삭제가 자주 필요한 경우
3. 메모리 공간을 연속적으로 확보하기 힘든 경우
```

```python
03) 헤드
- 연결리스트의 시작점을 가리키는 포인터(주소저장소)
- 첫번째 노드를 가리키는 속성(변수)
- 연결리스트를 탐색할 때는 항상 head부터 출발
-> head가 없다면 연결리스트 전체를 알 수 없음
```

```python
[체리, 바나나, 수박]
head -> [체리 | 바나나의 주소]-> [바나나 | 수박의 주소]-> [수박 | None]
```

```python
"""
체리,바나나, 수박 노드를 만들고 연결하기
""""
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

node1 = Node('체리')
node2 = Node('바나나')
node3 = Node('수박')
node1.next = node2 # node2에 연결
node2.next = node3 # node3에 연결

# head 설정: 첫번째 노드를 가리키게 함
head = node1 # head는 node1만 가리키므로 불변!

current = head # 한칸한칸 이동할 때 사용하는 변수
while current: # 값이 들어있다면 수행하라는 의미임
    print(current.data)
    current = current.next # 업데이트 코드 (업데이트하지 않으면 계속 node1 이기에 node1 -> node2 -> node3로 업데이트)
```

```python
04) 노드 삽입
- 연결리스트의 원하는 위치에 새로운 노드를 추가하여 기존 노드들과 포인터로 연결하는 동작
- 연결리스트는 연속된 메모리가 필요 없기 때문에 중간, 앞,뒤 어디든지 삽입 가능
```

```python
[이전노드] -> [다음노드]
[이전노드] -> [새노드] -> [다음노드]
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

node1 = Node('체리')  # [체리 | None]
node2 = Node('바나나')
node3 = Node('수박')
node1.next = node2 # node2에 연결
node2.next = node3 # node3에 연결

# head 설정: 첫번째 노드를 가리키게 함
head = node1 # head는 node1만 가리키므로 불변!

# 1. 맨 앞에 노드 삽입(딸기)
# [딸기] -> [체리] -> [바나나] -> [수박]

new_node = Node('딸기')
new_node.next = head # node1을 가리킴
head = new_node

# [문제 1]
# 1.2 맨 앞 노드 삽입(블루베리)
# [블루베리] -> [딸기] -> [체리] -> [바나나] -> [수박]
new_node2 = Node('블루베리')
new_node2.next = head # new_node딸기의 주소값을 new_node2의 next에 저장
head = new_node2 # head는 다시 new_node2를 가리킴

# 2. 중간노드 삽입(오렌지) 
# [블루베리] -> [딸기] -> [체리] -> [바나나] -> [오렌지] ->  [수박]

new_node3 = Node('오렌지')
target = node2 # 새로운 변수로 바나나 노드와 연결
new_node3.next = node3 # 오렌지 노드의 next를 수박의 주소값으로 연결
target.next = new_node3 # 바나나 노드가 오렌지 노드를 가리키도록 연결
# 순서가 매우 중요 
# 1. 새로운 노드가 그 다음 노드를 가리키도록, 
# 2. 이전 노드가 새로운 노드를 가리키도록


# [문제2]
# 2.2 중간노드 삽입(사과 ) 
# [블루베리] -> [딸기] -> [사과] -> [체리] -> [바나나] -> [오렌지] ->  [수박]

new_node4 = Node('사과')
target2 = new_node
new_node4.next = target2.next
target2.next = new_node4

# [문제3]
# 3. 마지막 노드 삽입(망고 ) 
# [블루베리] -> [딸기] -> [사과] -> [체리] -> [바나나] -> [오렌지] ->  [수박] -> [망고]

new_node5 = Node('망고')
target = node3
target.next = new_node5 # 마지막 노드이므로 주소 연결 할필요 없음



current = head # 한칸한칸 이동할 때 사용하는 변수
while current: # 값이 들어있다면 수행하라는 의미임
    print(current.data)
    current = current.next # 업데이트 코드 (업데이트하지 않으면 계속 node1 이기에 node1 -> node2 -> node3로 업데이트)
```

```python
06) 노드 클래스와 단일연결리스트 클래스 생성하기
```

```python
[사용할 메서드]
append() : 리스트의 맨 끝에 요소를 추가
prepend(): 리스트의 맨 앞에 요소를 추가
insert(): 리스트의 중간에 요소를 삽입
delete(): 리스트에서 특정 요소를 삭제
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class SinglyLinkedList:
    def __init__(self):
        self.head = None
    # 맨 뒤에 요소를 추가하는 작업
    def append(self, data):
        new_node = Node(data) # Node클래스를 이용해서 노드를 생성해 
        if not self.head:
            self.head = new_node # head는 new_node를 가리킴
            return
        last_node = self.head
        while last_node.next: # 마지막 노드를 찾아주는 코드
            last_node = last_node.next # 한칸 뒤로 움직임
        last_node.next = new_node # 1번과 2번 연결

    # 맨 앞에 요소를 추가하는 메서드
    def prepend(self, data):
        new_node = Node(data) # self. 안붙여서 따로 저장은 x , just head에 주소만 저장
        new_node.next = self.head  # 같은 클래스 안의 속성이면 앞에 self. 붙이기
        self.head = new_node # 새로운 노드가 head 이다

        
    # 중간 삽입 메서드(insert) # 어디에 값을 넣을지 index를 받아서 활용
    def insert(self, index, data):
        if index == 0: # 인덱스가 0이면 앞애 삽입
            self.prepend(data) # prepend 메서드 활용한다
            return # 종료코드
        new_node = Node(data) # 아니면 노드 생성
        current_node = self.head
        current_position = 0
        while current_node and current_position < index -1: 
            current_node = current_node.next # current를 한칸 이동
            current_position += 1
        if current_node is None: # 못찾은 경우 current_node = None
            print('Not found')
            return
        new_node.next = current_node.next # 새노드의 주소 -> 다음 노드의 주소를 저장
        current_node.next = new_node # 이전노드의 주소 -> 새노드의 주소
        
     # 삭제   
    def delete(self, key):
        if not self.head:
            print('List is empty')
            return
        if self.head.data == key: # 제거하고자 한 값이 head의 데이터와 같다면
            temp = self.head # head의 주소값을 temp에 저장
            self.head = self.head.next
            del temp
            return
        current_node = self.head
        previous_node = None
        while current_node and current_node.data != key:
            previous_node = current_node
            current_node = current_node.next
        if current_node is None:
            print(f'Node with data {key} not found')
            return
        previous_node.next = current_node.next
        del current_node
        
    def print_list(self):
        current_node = self.head
        while current_node:
            print(current_node.data, end = ' ->  ')
            current_node = current_node.next
        print('None')
            
            

sll = SinglyLinkedList() # def __init__(self) 만들어져서 객체 sll에 속성 head에 None 값 저장
sll.append(1) # 해당 클래스의 메서드 append의 data=1 전달
sll.append(2)
sll.append(3)
sll.append(4)
sll.print_list()
sll.prepend(0)
sll.print_list()
sll.insert(4, 3.5)
sll.print_list()
```
