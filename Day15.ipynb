```python
hospital = data['TbHospitalInfo']['row']
hospital
```

```python
information_ls = [] # 병원 정보를 담을 빈 리스트

for i in hospital:
    information_dict = {} # 각 병원에 대해 새로운 딕셔너리 만든다
    information_dict['기관명'] = i['DUTYNAME'] #기관명 : 병원의 이름 (키:밸류)
    information_dict['병원분류명'] = i['DUTYDIVNAM'] # 병원분류명
    information_dict['간이약도'] = i['DUTYMAPIMG']
    information_dict['주소'] = i['DUTYADDR']
    information_dict['전화번호'] = i['DUTYTEL1']
    information_ls.append(information_dict) # 반복되면서 마지막 부분만 남는것 방지 위함 -> 어딘가에 정보들 저장해놓기
information_ls
```
