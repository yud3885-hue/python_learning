```python
04) Selenium 라이브러 구조 
Selenium 라이브러리
   -> webdriver(모듈)
        -> Chrome(클래스)
            -.get(url) -> 메서드 : 페이지 이동
            -.find_element()-> 메서드: 요소 1개찾기
            -.find_elements()-> 메서드: 요소 여러개 찾기
            -.page_source-> 속성: HTML소스
            -.current_url-> 속성: 현재URL
            -.title-> 속성: 페이지 제목
            -.quit()-> 메서드: 브라우저 종료
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
#selenium라이브러리
#webdriver모듈(하위패키지)
#common 모듈(하위패키지)
#by 모듈(파일)
#By 클래스 -> 요소를 찾는 기준(ID, CLASS_NAME 등)
import time 

#1. 브라우저 실행 
driver = webdriver.Chrome() #객체 생성 - 빈브라우저 창만 뜬 상태 
#2.  사이트 접속
driver.get("https://example.com") #페이지 이동 
#3. JavaScript 실행 대기 
time.sleep(2) 
#4.완성된 HTML가져오기
print(driver.page_source)
print(driver.title)

driver.quit() #사용 후 브라우저 종료
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import os
#운영체제의 줄임말(Operating System)
# 컴퓨터의 파일/폴더 경로, 디렉토리 조작, 환경변수 등을 다루는 기능들을 모아둔 모듈

html = """
<div class = "content">
    <h2>게시판</h2>
    <ul>
        <li>공지사항</li>
        <li>자유게시판</li>
        <li>질문게시판
            <ul>
                <li>답변대기</li>
                <li>답변완료</li>
            </ul>
        </li
    </ul>
    <button id = "loginBtn">로그인</button>
<div>
"""
with open("test.html", "w", encoding = "utf-8") as f:
    f.write(html)
driver = webdriver.Chrome()
path = os.path.abspath("test.html")#절대경로 구하기 
file_url = "file:///" +path #브라우저가 읽을 수 있는 형식으로 변환
driver.get(file_url)
#1. div.content
box = driver.find_element(By.CSS_SELECTOR, "div.content")
print("박스: ", box.text.strip()) #불필요한 문자 제거 
print("-" * 10)
#2. #loginBtn
btn = driver.find_element(By.CSS_SELECTOR, "#loginBtn")
print("버튼: ", btn.text)
print("-" * 10)
#3. ul>li
menus = driver.find_elements(By.CSS_SELECTOR, "ul > li")
for meny in menus:
    print("메뉴: ", menu.text)
print(len(menus))
print("-" *10)

#4. div.content바로 아래 ul의 직계자식 li만 정확히 선택 
menus = driver.find_elements(By.CSS_SELECTOR, "div.content > ul > li")
for menu in menus:
    print("메뉴: ", menu.text)
print(len(menus))
print("-"*10)

#5. li모두 가져오기 
li = driver.find_elements(By.CSS_SELECTOR, "li")
for item in li:
    print(item.text)
print(len(li))

driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By 
from selenium.webdriver.common.keys import Keys
import time

driver = webdriver.Chrome()
#1. 로그인 페이지 열기 
driver.get("http://the-internet.herokuapp.com/login")
#2. 아이디입력
username = driver.find_element(By.ID, "username")
username.send_keys("tomsmith")
#3. 비밀번호 입력
password = driver.find_element(By.ID, "password")
password.send_keys("SuperSecretPassword!")
#4. 로그인 버튼 클릭
login_button = driver.find_element(By.CLASS_NAME, "radius")
login_button.click()
#5. 결과 확인을 위해 잠시 대기
time.sleep(2)

#6. 스크린샷 저장(로그인 후 화면)
driver.save_screenshot("login_result.png")
print("로그인 테스트 완료!")
driver.quit()
```

## 03 | 무한스크롤 5번 내리기

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time

driver = webdriver.Chrome()
driver.get("https://the-internet.herokuapp.com/infinite_scroll")
time.sleep(2)
#스크롤 5번 내리기 
for i in range(5):
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    #.scrollTo(x,y): 브라우저 화면을 좌표(x,y)로 이동해주는 기능
    #y축좌표: 페이지의 전체높이-> 이값을 입력하면 페이지의 가장 아래로 이동!
    print(f"스크롤{i+1}회 완료")
    time.sleep(2)
print("무한스크롤 테스트 완료")
driver.quit()
```
