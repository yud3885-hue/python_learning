```python
06) 집합연산 
- 교집합: A에도 속하고 B에도 속하는 원소들을 골라놓은 집합
- 합집합: A에 속하거나 B에 속하는 원소들로 이루어진 집합
- 차집합: 한집합에서 다른 집합 요소를 뺀 새로운 집합을 생성하는 집합 연산
```

```python
setA = {10,20,30,40,50,60}
setB = {50,60,70,80,90,100}
print(setA & setB)
print(setA | setB)
#차집합
print(setA -setB)
print(setB -setA)
```

```python
scores = {"Alice": 85, "Bob": 92, "Charlie": 78}
print(scores.get("Bob"))

scores["David"] = 90 
print(scores)

s = {"David": 90}
scores.update(s)
print(scores)

for key,values in scores.items():
    print(f'key: {key}, values: {values}')
    
```
