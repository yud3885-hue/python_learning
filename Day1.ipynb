## 추가 01 | 문자열 포매팅

```python
#추가 | 문자열 포매팅
- 문자열 안에 변수나 값을 넣는 방법
01) format() 메서드
- format메서드는 문자열의 중괄호{} 안에 특정 값을 삽입하는데 사용
```

```python
[작성법]
"문자열 ~~{}".format('특정값')  # ()안의 특정값을 문자열의 {} 안에 넣어주는 기능
```

```python
'''
1. 기본 순서 대입 방식

중괄호 {}가 나타나는 순서대로 format() 내부의 값이 차례차례 삽입됨.

작성 예: '{}와 {}'.format('사과', '바나나')

2. 인덱스(숫자) 지정 방식

중괄호 안에 0부터 시작하는 숫자를 넣어, 삽입할 값의 순서를 직접 지정할 수 있음.

작성 예: '{1} 먼저, 그다음 {0}'.format('첫번째값', '두번째값')

3. 키워드(이름) 지정 방식

중괄호 안에 직관적인 이름을 넣고, format() 안에서 해당 이름에 값을 할당하는 방식임.

작성 예: '{name}의 나이는 {age}살'.format(name='김철수', age=25)
'''
```

```python
print("My name is {}".format('James'))
```

```python
name = "J"
print("My name is {}".format(name))
```

```python
print("My name is {0}. I'm {1}years old.".format("James", 25))
```

```python
print("My name is {name}. {age}".format(name='james', age =25))
```

```python
text = "My name is {}"
print(text.format("James"))
```
