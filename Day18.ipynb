```python
class TreeNode:
    def __init__(self,key):
        self.key = key
        self.left = None
        self.right = None
class Tree:
    def __init__(self):
        self.root = None
    def insert(self, key):
        if self.root is None:
            self.root = TreeNode(key)
        else:
            self._insert_level_order(self.root, key) 
    def _insert_level_order(self, node,key):
        queue = [node]
        while queue:
            current = queue.pop(0) # 탐색하기 위한 준비
            if current.left is None:
                current.left = TreeNode(key)
                return
            else:
                queue.append(current.left)
            if current.right is None:
                current.right = TreeNode(key)
                return
            else:
                queue.append(current.right)
    def display(self):
        lines, *_ = self.display_recursive(self.root)
        for line in lines:
            print(line)
    def display_recursive(self,node):
        if node is None:
            return [], 0, 0
        left_lines, left_width, left_height = self.display_recursive(node.left)
        right_lines, right_width, right_height = self.display_recursive(node.right)
        node_key_width = len(str(node.key))
        total_width = left_width+ node_key_width + right_width
        total_height = max(left_height, right_height) + 1
        while len(left_lines) < total_height:
            left_lines.append(" "* left_width)
        while len(right_lines) < total_width:
            right_lines.append(" "* left_width)
        middle_line = " " * left_width +str(node.key) + " " * right_width
        lines = [middle_line]
        for i in range(total_height -1 ):
            lines.append(left_lines[i] + " " * node_key_width + right_lines[i])
        return lines, total_width, total_height
bt = Tree()
arr = [1,2,3,4,8,6,7]
for a in arr:
    bt.insert(a)
bt.display()
```

# 03 | 이진탐색트리

```python
01) 이진탐색트리란?
- 이진트리에서 추가적인 규칙이 붙은것
-> 왼쪽 자식노드 < 부모 노드 < 오른쪽 자식 노드
* 부모 노드보다 왼쪽 자식 노드가 작음
* 부모 노드보다 오른쪽 자식 노드가 크다
```

```python
02) 이진탐색 트리의 특징
1. 구조가 단순
- 각 노드는 최대 2개의 자식 노드를 가짐
2. 이진탐색을 통해 아주 빠르게 검색 가능
3. 노드 삽입이 쉽다
4. 중위순회를 통해 노드값을 오름차순으로 얻을 수 있음
```

```python
03) 노드 탐색
- 찾고자 하는 키값을 현재 노드의 값과 비교
- 찾고자 하는 키값과 현재 노드의 값이 같으면 탐색 종료
- 찾고자 하는 키값이 현재 노드보다 작으면 왼쪽 서브트리로 이동
                                   크면 오른쪽 서브트리로 이동
```

```python
04) 이중탐색 트리 구현
[사용되는 클래스]
1. class TreeNode: 노드 객체를 정의
2. class BinaryTree: 이진 트리 전체를 표현하는 클래스
[사용되는 메서드]
1. def insert(self,key): 새로운 데이터를 트리에 삽입하기 시작하는 메서드
2. def _insert_recursive(self, node, key): 값을 비교하면서 삽입할 위치 찾기
```

```python
class TreeNode:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None
        

class BinaryTree:
    def __init__(self):
        self.root = None
    def insert(self, key):
        if not self.root: # 비어있으면 수행해
            self.root = TreeNode(key)
        else:
            self._insert_recursive(self.root, key)
    def _insert_recursive(self, node, key):
        # 새로 추가하는 값이 부모노드보다 작으면 왼쪽으로 추가 # 핵심부분!!!
        if key < node.key:
            if node.left is None:
                node.left = TreeNode(key)
            else:
                self._insert_recursive(node.left, key)
        # 새로 추가하는 값이 부모노드보다 크면 오른쪽으로 추가
        else:
            if node.right is None:
                node.right = TreeNode(key)
            else:
                self._insert_recursive(node.right, key)
                
    def display(self):
        lines, *_ = self.display_recursive(self.root)
        for line in lines:
            print(line)
    def display_recursive(self,node):
        if node is None:
            return [], 0, 0
        left_lines, left_width, left_height = self.display_recursive(node.left)
        right_lines, right_width, right_height = self.display_recursive(node.right)
        node_key_width = len(str(node.key))
        total_width = left_width+ node_key_width + right_width
        total_height = max(left_height, right_height) + 1
        while len(left_lines) < total_height:
            left_lines.append(" "* left_width)
        while len(right_lines) < total_width:
            right_lines.append(" "* left_width)
        middle_line = " " * left_width +str(node.key) + " " * right_width
        lines = [middle_line]
        for i in range(total_height -1 ):
            lines.append(left_lines[i] + " " * node_key_width + right_lines[i])
        return lines, total_width, total_height
bt = BinaryTree()
arr = [50, 30, 70, 20, 40, 60, 80]
for a in arr:
    bt.insert(a)
bt.display()\\#  한번 더보기
```

```python
(암기) -> 루트를 언제 방문하냐가 기준
04 | 이진트리 순회
-> 깊이 우선 탐색
1. 전위순회(preorder traverse)
01) 루트 방문
02) 왼쪽 서브트리 방문
03) 오른쪽 서브트리 방문
방문순서: 50 - 30 - 20 - 40 - 70 - 60 - 80

2. 중위순회(inorder traverse)
01) 왼쪽 서브트리 방문
02) 루트 방문
03) 오른쪽 서브트리 방문

3. 후위순회(postorder traverse)
01) 왼쪽 서브트리 방문
02) 오른쪽 서브트리 방문
03) 루트 방문
```

```python
      50      
  30      70  
20  40  60  80 

1. 전위순회: 50-30-20-40-70-60-80 
2. 중위순회: 20-30-40-50-60-70-80 (오름차순)
3. 후위순회: 20-40-30-60-80-70-50
```

```python
class BinaryTreeTraversal(BinaryTree): # 상속받음
    def preorder_traversal(self, root):
        if root:
            print(root.key, end = ' ' )
            self.preorder_traversal(root.left) # 재귀호출
            self.preorder_traversal(root.right)
            
    def inorder_traversal(self, root): # 중위순회(왼쪽 -> 루트 -> 오른쪽)
        if root:
            self.inorder_traversal(root.left) # 20까지 파고들도록 재귀호출
            print(root.key, end = ' ') # 20 방문했다고 노드 출력
            self.inorder_traversal(root.right) 
            
            
    def postorder_trversal(self, root): # 후위순회 (왼쪽-> 오른쪽 -> 루트)
        if root:
            self.postorder_trversal(root.left)
            self.postorder_trversal(root.right)
            print(root.key, end = ' ')
            
arr = [50,30,70,20,40,60,80]
btt = BinaryTreeTraversal()
for a in arr:
    btt.insert(a) #  상속받은 클래스의 메서드 사용 가능
btt.display()
btt.preorder_traversal(btt.root) # 50-30-20-40-70-60-80
print()
btt.inorder_traversal(btt.root)
print()
btt.postorder_trversal(btt.root)

# 재귀호출 그림 따라그리기
# 이 코드는 진짜 개중요함
```
